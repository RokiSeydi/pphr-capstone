# Question 5: (5 Marks)
Below is a class called `CardioRiskProfile` which stores information about a patient's risk for cardiovascular disease, and has the ability to assess their risk of disease with the `assess_risk` function. 

Your task is to develop code to make three instances of the class with dummy data that causes the assess_risk method to return 
1. one `Low Risk`
1. one `Moderate Risk`
1. one `High Risk`
1. print the output

````python
# Starter code

# CardioRiskProfile 
class CardioRiskProfile:
    def __init__(self, patient_name, systolic_bp, diastolic_bp, regular_exercise, diet_type):
        """
        Initialize a patient's cardiovascular risk profile.
        
        :param patient_name: Name of the patient (str)
        :param systolic_bp: Systolic blood pressure (int)
        :param diastolic_bp: Diastolic blood pressure (int)
        :param regular_exercise: True if patient exercises regularly (bool)
        :param diet_type: Patient's diet ("vegan", "vegetarian", "omnivore")
        """
        self.patient_name = patient_name
        self.systolic_bp = systolic_bp
        self.diastolic_bp = diastolic_bp
        self.regular_exercise = regular_exercise
        self.diet_type = diet_type

    def assess_risk(self):
        """
        Assesses the patient's cardiovascular risk level based on predefined criteria.
        
        :return: A string indicating the risk level ("High Risk", "Moderate Risk", "Low Risk")
        """
        # Check for High Risk criteria first
        if self.systolic_bp >= 140 or self.diastolic_bp >= 90:
            return "High Risk"
        
        # Check for Low Risk criteria
        elif self.systolic_bp < 120 and self.diastolic_bp < 80 and \
             self.regular_exercise and (self.diet_type == "vegan" or self.diet_type == "vegetarian"):
            return "Low Risk"
        
        # Any other combination is considered Moderate Risk.
        else:
            return "Moderate Risk"
````
