import numpy as np

````python
# Starter code
import numpy as np

def mystery_function(values, threshold):
    flagged = values >= threshold
    n = np.sum(flagged)
    if n == 0:
        return None
    return float(np.mean(values[flagged]))

# Example usage 1
v = np.array([0.9, 1.4, 2.2, 5.1, 3.3])
print(mystery_function(v, 2.0))    # Output: 3.5333333333333337

# Example usage 2
print(mystery_function(v, 9.0))    # Output: None
````
